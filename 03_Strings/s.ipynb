## Strings

```python
str = "12345678910"
print(str[0:10:2])
```

```python
s1 = "Masala Chai"
print(s1.lower())
print(s1.upper())
print(s1)
# s1 remainis the same. this proves that strings are immutable
```

## Stripping

used in web development to remove whitespace from the beginning and end of a string.

```python
s2 = "   ved   "
print(s2)
print(s2.strip())
```

## Replace the string

```python
print(s1.replace("Masala", "Lemon"))
# Masala got replaced by lemon
print(s1.replace("Masala Chai", "Lemon Tea"))
```

## Convert a string to a list

```python
st = "Ginger, Lemon, Honey, Mint"
print(st.split())
print(st.split(", "))
```

## Convert list to a string

```python
list = ["lemon", "Giinger", "Mint"]

print("".join(list))
print(", ".join(list))
print("-".join(list))
```

## Placeholder  

```python
chai = "Masala"
quantity = 2
stri = "i have ordered {} cups of {} chai"
print(stri.format(chai, quantity))
```
