## Lists

# Slicing and everything is same as strings. Lists are mutable, meaning you can change their content without changing their identity. Lists can contain items of different types, but usually the items all have the same type.    

```python
l1 = [1,2,3]
l2=[4,5,6]
l3=l1+l2
print(l3)
```

```python
variety = ["apple", "banana", "cherry", "kiwi", "mango"]
variety[1:1] = "plum"
print(variety)
variety[1:1] = ["plum"]
print(variety)
variety[1:3] = []
print(variety)
variety[1:4] =[]
```

## Some loops

```python
for fruit in variety:
    print(fruit)

for fruit in variety:
    print(fruit, end=("-"))
```

Some Methods

```python
variety = ["apple", "banana", "cherry", "kiwi", "mango"]
variety.append("Grapes")
print(variety)

variety[4] = "Grapes"
variety[5] = "Mango"
variety.pop()
print(variety)

variety.remove("banana")
print(variety)

variety.insert(2, "Mango")
print(variety)
```

## Fancy way to decalre a list

```python
sq_num = [x**2 for x in range(10)]
print(sq_num)

cube_num = [x**3 for x in range(7)]
print(cube_num)

num= [(x+5) for x in range(7)]
print(num)
```
