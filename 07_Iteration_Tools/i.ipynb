## Iteration Tools

```python
f = open('hi.py')
print(f.readline())
print(f.readline())
print(f.readline())
print(f.readline())

# this proves that the e,pty lines are also being counted in the .readline() method
```

```python
g = open('hi.py')
print(g.__next__())
print(g.__next__())
print(g.__next__())
print(g.__next__())
print(g.__next__())
#  this is the raw method which means how python does tings internally
```

```python
myList = [1,2,3,4]

I = iter(myList)
print(I.__next__())
print(I.__next__())
print(I.__next__())
print(I.__next__())
print(I.__next__())
```
