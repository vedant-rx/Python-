## Dictionaries

key -> value

```python
chai = {"masala":"spicy" , "ginger":"zesty", "green":"Mild"}
print(chai["masala"])

chai["green"] = "Fresh"
print(chai["green"])
```

```python
sq = {x:x**2 for x in range(6)}
print(sq)
```

```python
key = ["1","2","3"]
value = "one"

dic = dict.fromkeys(key, value)
print(dic)

dic = dict.fromkeys(key, key)
print(dic)
```
