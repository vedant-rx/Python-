## Basic class and Object(Create a car class with attributes like brand and model. then create an instance of the class)

```python
class Car:
    def __init__(self, brand, model):
        self.brand = brand
        self.model = model


my_Car = Car("Toyota", "Altis")
print(my_Car.brand)
print(my_Car.model)

my_2Car = Car("Suzuki", "Baleno")
print(my_2Car.brand)
print(my_2Car.model)
```

## Adding a method

```python
class Car:
    def __init__(self, brand, model):
        self.brand = brand
        self.model = model
    def fullname(self):
        return f"{self.brand} {self.model}"


my_Car = Car("Toyota", "Altis")
print(my_Car.brand)
print(my_Car.model)
print(my_Car.fullname())

my_2Car = Car("Suzuki", "Baleno")
print(my_2Car.brand)
print(my_2Car.model)
print(my_2Car.fullname())
```

## Create an electric car class that inherits from the car class and adds an attribute for battery capacity

```python
class Electric_Car(Car):
    def __init__(self, brand, model, battery_size):
        super().__init__(brand, model)
        self.battery_size = battery_size

tesla = Electric_Car("Tesla", "123", "15kgw")

print(tesla.battery_size)
print(tesla.model)
print(tesla.brand)
print(tesla.fullname())
```

## Encapsulation 
modify the car class to encapsulate the brand attribute, making it private and provide a getter method to access it.

```python
class Car:
    def __init__(self, brand, model):
        self.__brand = brand
        self.model = model

    def get_brand(self): #this is the standard way to create a getter
        return self.__brand + "!"
    def fullname(self):
        return f"{self.__brand} {self.model}"


my_Car = Car("Toyota", "Altis")
print(my_Car.get_brand())
print(my_Car.model)
print(my_Car.fullname())

my_2Car = Car("Suzuki", "Baleno")
print(my_2Car.get_brand())
print(my_2Car.model)
print(my_2Car.fullname())
```

## Polymorphism
demonstrate polymorphism by defining a method fuel_type in both electric car and car class but with different behaviour.

```python
class Car:
    def __init__(self, brand, model):
        self.__brand = brand
        self.model = model

    def get_brand(self): #this is the standard way to create a getter
        return self.__brand + "!"
    def fullname(self):
        return f"{self.__brand} {self.model}"
    def fuel_type(self):
        return "petrol or diesel"


my_Car = Car("Toyota", "Altis")
print(my_Car.get_brand())
print(my_Car.model)
print(my_Car.fullname())
print(my_Car.fuel_type())

my_2Car = Car("Suzuki", "Baleno")
print(my_2Car.get_brand())
print(my_2Car.model)
print(my_2Car.fullname())
print(my_2Car.fuel_type())


class Electric_Car(Car):
    def __init__(self, brand, model, battery_size):
        super().__init__(brand, model)
        self.battery_size = battery_size
    def fuel_type(self):
        return "Electric charge"

tesla = Electric_Car("Tesla", "123", "15kgw")

print(tesla.battery_size)
print(tesla.model)
print(tesla.get_brand())
print(tesla.fullname())
print(tesla.fuel_type())

```

## Class Variables

## Static Methods
add a static method to the car class that return a general description of what a car is.

```python
class Car:
    total = 0
    def __init__(self, brand, model):
        self.__brand = brand
        self.model = model
        Car.total = Car.total + 1

    def get_brand(self): #this is the standard way to create a getter
        return self.__brand + "!"

    def fullname(self):
        return f"{self.__brand} {self.model}"

    def fuel_type(self):
        return "petrol or diesel"

    @staticmethod
    def general_des():
        return "this is not ev car"

class Electric_Car(Car):
    def __init__(self, brand, model, battery_size):
        super().__init__(brand, model)
        self.battery_size = battery_size
    def fuel_type(self):
        return "Electric charge"

my_Car = Car("Toyota", "Altis")
print(my_Car.get_brand())
print(my_Car.model)
print(my_Car.fullname())
print(my_Car.fuel_type())
print(my_Car.general_des())


my_2Car = Car("Suzuki", "Baleno")
print(my_2Car.get_brand())
print(my_2Car.model)
print(my_2Car.fullname())
print(my_2Car.fuel_type())
print(my_2Car.general_des())




tesla = Electric_Car("Tesla", "123", "15kgw")

print(tesla.battery_size)
print(tesla.model)
print(tesla.get_brand())
print(tesla.fullname())
print(tesla.fuel_type())

print(Car.total)
print(Car.general_des())

```

## Multiple Inheritance
create two classes battery and enigne and let the electric car class inherit from both of them, demonstrating multiple inheritance.
