## Scope

```python
user = "vedant" # this is global variable

def func():
    user = "vedx" # this is local variable
    print(user)

func()
print(user)
func()
```

```python
x = 5
def fun2(y):
    print(x+y) 

fun2(5)

## when no local variables are available the it takes the global vlaue
```

```python
def chaicoder(num):
    def actual(x):
        return x ** num
    return actual

f = chaicoder(2)
g = chaicoder(3)

print(f)
print(g)

print(f(5))
print(g(6))
```
