```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Data structure: [restaurant_id, 2021, 2022, 2023, 2024]

sales_data = np.array([
    [1, 150000, 180000, 220000, 250000],  # Paradise Biryani
    [2, 120000, 140000, 160000, 190000],  # Beijing Bites
    [3, 200000, 230000, 260000, 300000],  # Pizza Hub
    [4, 180000, 210000, 240000, 270000],  # Burger Point
    [5, 160000, 185000, 205000, 230000]   # Chai Point
])

print("=== zomato Sales analysis")
print("Sales data shape", sales_data.shape)
print("sample data for 1st three restaurants", sales_data[:3])
```

```python
# total sales per year
print(np.sum(sales_data, axis = 0))
yearly_total = np.sum(sales_data[:, 1:], axis=0)
print(yearly_total)

#minimum sales per rest
min_sale = np.min(sales_data[:, 1:], axis=1)
print(min_sale)

#Maximum sale sper year
max_sales = np.max(sales_data[:, 1:], axis=0)
print(max_sales)

#Average sales per rest
avg = np.mean(sales_data[:, 1:], axis=1)
print(avg)
```

```python
#Cumulative Sales of a rest
cum = np.cumulative_sum(sales_data[:, 1:], axis =1)
print(cum)

plt.figure(figsize=(10,6))
plt.plot(np.mean(cum, axis =0))
plt.xlabel = "Years"
plt.ylabel = "Sales"
plt.title = "Mean Cumulative sales"
plt.grid(True)
plt.show
```

```python
vector1 = np.array([1,2,3,4,5])
vector2 = np.array([6,7,8,9,10])

print("vector sum is : ", vector1 + vector2)
print("vector multoiplication is : ", vector1 * vector2)

print("Dot Product is : ", np.dot(vector1, vector2))
angle = np.arccos(np.dot(vector1, vector2) / (np.linalg.norm(vector1) * np.linalg.norm(vector2)))
print(angle)
```
