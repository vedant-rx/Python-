```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
array1 = np.array([[1,2,3],[4,5,6]])
array2 = np.random.random((3,3))
array3 = np.full((2,3),4)

np.save('Array 1', array1)
np.save('Array 2', array2)
np.save('Array 3', array3)

loaded1 = np.load('Array 1.npy')
loaded2 = np.load('Array 2.npy')
loaded3 = np.load('Array 3.npy')
print(loaded1)
print(loaded2)
print(loaded3)
```

```python
try:
    logo = np.load('numpy-logo.npy')
    plt.figure(figsize=(10,5))
    plt.subplot(121)
    plt.imshow(logo)
    plt.title("Normal Image")
    plt.grid(True)

    darl_logo = 1 - logo
    plt.subplot(122)
    plt.imshow(darl_logo)
    plt.title("Dark Image")
    plt.grid(True)
except FileNotFoundError:
    print("File not FOumd")
```
