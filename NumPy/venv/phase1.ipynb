## Numpy basics and arrays


```python
import numpy as np
```

## creating a numpy array

```python
arr_1d=np.array([1,2,3])
print("the one dimen array is",arr_1d)

arr_2d=np.array([[1,2],[2,3]])
print("the two dimen array is",arr_2d)
```

## List vs Numpy Array

```python
py_list=[1,2,3,4,5]
print("python list multi",py_list*2)

np_array=np.array([1,2,3,4,5])
print("python array multi",np_array*2)

import time
start=time.time()
py_list=[i*2 for i in range(100000)]
print("list time",time.time() - start)

start=time.time()
np_array=np.arange(100000) * 2
print("array time",time.time() - start)
```

## creating arrays from scratch

```python
zeros=np.zeros((3,4))
print("zero array",zeros)

ones=np.ones((1,2))
print("ones array",ones)

full=np.full((2,2),5)
print("full array",full)

random=np.random.random((2,2))
print("random array",random)

seq=np.arange(0,11,2)
print("seq array",seq)


```

## vectors and matrices and tensor

```python
vector = np.array([1,2,3])
print("vector is ",vector)

matrix = np.array([[1,2],[2,3]])
print("matrix is ",matrix)

tensor = np.array([[[1,2],[3,4]],
                   [[5,6],[7,8]]])
print("tensor is ",tensor)
```

## Array Properties

```python
arr = np.array([[1,2,3],[4,5,6]])
print("the shape is:",arr.shape) 
print("the dimesnion is:",arr.ndim) 
print("the size is:",arr.size) 
print("the Dtype is:",arr.dtype) 
```

## Array Reshaping

```python
original = np.arange(12)
print("original array is ",original)

reshaped = original.reshape((3,4))
print("reshaped array is ",reshaped)

flattened = reshaped.flatten()
print("flattened array is ",flattened)

## returns view, does not copy
raveled = reshaped.ravel()
print("RAVELED array is ",raveled)

transpose = reshaped.T
print("transposed array is ",transpose)
```
