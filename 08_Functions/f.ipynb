## Functions Definitions

1. Write a function to find the square of a number and retur the same

```python
def vec(num):
    return num*num
print(vec(2))
print(vec(7))
```

```python
def su(one, two):
    return one + two
print(su(30,2))
print(su(30,20282))
```

```python
def mul(on, tw):
    return on * tw
print(mul(2,5))
print(mul(2,"ved"))
print(mul("ved", 4))
```

```python
import math
def cir(rad):
    circum = 2 * math.pi * rad
    area = math.pi * rad**2
    return circum, area 
print(cir(5))
```

## Args Functions

```python
def hag(*args): ## we can write anythign instead of args but it is not a good practice
    print(args)
    print(*args)

    return sum(args)

print(hag(1,3,2))
print(hag(1,3,2,4,5))
print(hag(1,3,2,4,5,6,7))
print("\n")
def h(*args):
    for i in args:
        print(i*2)
print(h(1,2,3))
```

## Kwargs Functions

used for key-value pairs

```python
def kw(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")

kw(name = "shaktiman", power = "hero")
kw(name = "shaktiman", power = "hero", doctor = "jackal")
kw(name = "shaktiman")
```

## Generator Function 

```python
def gen(limit):
    for i in range(2, limit + 1, 2):
        yield i

for num in gen(10):
    print(num)
```

## Recursion Function

```python
def factorial(n):
    if n==0:
        return 1
    else:
        return n * factorial(n-1)
    
print(factorial(5))
```
